```python
# Re-applying column renaming from previous steps (Part F)
df = df.rename(columns={'fixed acidity': 'fixed_acidity', 'volatile acidity': 'volatile_acidity'})
```

#### 38. Create a new column named Acidity_Ratio.
Acidity_Ratio = fixed acidity / volatile acidity

```python
# Create Acidity_Ratio column
df['Acidity_Ratio'] = df['fixed_acidity'] / df['volatile_acidity']

# Display the first few rows with the new column
print(df[['fixed_acidity', 'volatile_acidity', 'Acidity_Ratio']].head())
```

#### 39. Create a new column named Alcohol_Level.
Alcohol \u2265 Mean Alcohol \u2192 High
Otherwise \u2192 Low

```python
# Calculate the mean alcohol level
mean_alcohol = df['alcohol'].mean()

# Create Alcohol_Level column
df['Alcohol_Level'] = df['alcohol'].apply(lambda x: 'High' if x >= mean_alcohol else 'Low')

# Display the first few rows with the new column and its value counts
print(df[['alcohol', 'Alcohol_Level']].head())
print('\nAlcohol_Level Value Counts:')
print(df['Alcohol_Level'].value_counts())
```

#### 40. Normalize the alcohol column using Min-Max Normalization

```python
from sklearn.preprocessing import MinMaxScaler

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Normalize the 'alcohol' column
df['alcohol_normalized'] = scaler.fit_transform(df[['alcohol']])

# Display the original and normalized alcohol columns
print(df[['alcohol', 'alcohol_normalized']].head())

# Display min and max of the normalized column to verify
print(f"\nMin of normalized alcohol: {df['alcohol_normalized'].min()}")
print(f"Max of normalized alcohol: {df['alcohol_normalized'].max()}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df['alcohol'], kde=True)
plt.title('Distribution of Alcohol Content')
plt.xlabel('Alcohol Content')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['quality'], bins=len(df['quality'].unique()), kde=False, color='skyblue')
plt.title('Distribution of Wine Quality')
plt.xlabel('Quality')
plt.ylabel('Frequency')
plt.xticks(sorted(df['quality'].unique()))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='alcohol', y='quality', data=df, hue='quality', palette='viridis', legend='full')
plt.title('Alcohol Content vs. Wine Quality')
plt.xlabel('Alcohol Content')
plt.ylabel('Quality')
plt.grid(linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='fixed_acidity', y='pH', data=df, hue='quality', palette='coolwarm', legend='full')
plt.title('pH vs. Fixed Acidity')
plt.xlabel('Fixed Acidity')
plt.ylabel('pH')
plt.grid(linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['alcohol'], color='lightcoral')
plt.title('Box Plot of Alcohol Content')
plt.ylabel('Alcohol Content')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['residual sugar'], color='lightgreen')
plt.title('Box Plot of Residual Sugar')
plt.ylabel('Residual Sugar')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
avg_alcohol_quality = df.groupby('quality')['alcohol'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='quality', y='alcohol', data=avg_alcohol_quality, palette='mako')
plt.title('Average Alcohol Content by Wine Quality')
plt.xlabel('Wine Quality')
plt.ylabel('Average Alcohol Content')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
correlation_matrix = df.corr(numeric_only=True)
quality_correlations = correlation_matrix['quality'].drop('quality') # Drop self-correlation
highest_correlation_feature = quality_correlations.abs().idxmax()
max_correlation_value = quality_correlations.loc[highest_correlation_feature]

print(f"Feature with highest correlation to Wine Quality: {highest_correlation_feature}")
print(f"Correlation value: {max_correlation_value:.2f}")
```
